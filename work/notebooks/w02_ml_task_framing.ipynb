## 5. Why ML beats a fixed rule here

A fixed rule could flag pages using one threshold, such as CTR below a chosen value, but that would ignore important context. CTR depends on factors such as search position, impressions, content type, age, and engagement. ML can combine several signals and learn a more flexible pattern for prioritization instead of relying on one arbitrary cutoff. The purpose is not to replace human review, but to make the review queue more useful.
